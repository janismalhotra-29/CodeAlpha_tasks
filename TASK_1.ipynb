```python
df=pd.read_csv("Messy_Employment_India_Dataset.csv")
```

```python
df.head()
```

```python
text_cols = ['Status','Age Group','Education','Industry','Location','AI Risk']

for col in text_cols:
    df[col] = df[col].str.strip().str.title()
```

```python
df.head()
```

```python
df['Age Group'] = df['Age Group'].str.replace('_', '-')
```

```python
for col in ['Status','Education','Industry','Location','AI Risk','Age Group']:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
df['Years of Experience'].fillna(df['Years of Experience'].median(), inplace=True)
```

```python
df['Monthly Salary (INR)'].fillna(df['Monthly Salary (INR)'].median(), inplace=True)
```

```python
df['Date Recorded'] = pd.to_datetime(df['Date Recorded'])
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.info()
df.isnull().sum()
```

```python
df.sample(10)
```

```python
df.to_csv("cleaned_employment_india.csv", index=False)
```
